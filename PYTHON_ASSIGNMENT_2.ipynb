```python
#LIST CREATION
#Create a list named age_list with five integer element
age_list=[24,25,26,27,28]
```

```python
#Create a list named name_list with five string elements
Name_list =['reshma','joy','basil','lucas','daisy']
```

```python
#Append the string "Yazhini" to name_list
Name_list.append("Yazhini")
```

```python
#find max age,min age,sum of all ages from age_list.
max_age=max(age_list)
min_age=min(age_list)
sum_age=sum(age_list)
print(max_age,min_age,sum_age)
```

DICTIONARY

```python
#creata a dictionary and print one specific mark
student_marks={
    "daisy":50,
    "lucas":95,
    "basil":45,
    "joy":32,
    "reshma":85
}
print(student_marks["reshma"])
```

```python
#add student name to dictionary
student_marks["janani"]=80
```

```python
#use the key(),values(),and items()
print(student_marks.keys())
print(student_marks.values())
print(student_marks.items())
```

SET OPERATIONS

```python
#create a set name as new_set
my_set ={'a','e','i','o','u','a','a','i'}
print(my_set)
```

```python
my_set.add('s')
```

```python
#create two set
set1={1,3,5,7,9}
set2={2,3,5,8,10}
```

```python
#COMPUTE AND PRINT THE UNION
union_set=set1.union(set2)
print(union_set)
```

```python
#intersection
intersection_set=set1.intersection(set2)
print(intersection_set)
```
